5. Create two random numpy array of size n x p and p x m and then compute the dot product

```python
n, p, m = 2, 3, 4

rand_arr_one = np.random.rand(n,p)
rand_arr_two = np.random.rand(p,m)

arr_product = np.dot(rand_arr_one, rand_arr_two) # can also use @ to multiply two matrix
print(arr_product)
```
